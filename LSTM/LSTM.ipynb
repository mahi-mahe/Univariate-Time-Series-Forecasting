An LSTM focused pipeline dedicated for hourly energy consumption forecasting. The reason behind using LSTM is derived from comparative study between ARIMA, Prophet, Naive, XGBoost and LSTM gives least MAE, RMSE and MAPE. An epoch represents one full pass of the entire training dataset through a machine learning model. It allows model to update weights, reducing error and improving accuracy over multiple iterations. Too few epoch cause underfitting, while too many epochs lead to overfitting.

```
"""
===================================================================
ELECTRICITY CONSUMPTION FORECASTING - LSTM-FOCUSED PIPELINE
Data-Driven Hourly Forecasting of Future Energy Demand
Author: Experienced ML Developer
Purpose: Production-ready LSTM model for accurate hourly energy demand prediction
===================================================================
"""

# ============================================================
# PHASE 0: IMPORTS & CONFIGURATION
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime, timedelta
import warnings
warnings.filterwarnings('ignore')

# Statistical & Time Series
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.tsa.stattools import adfuller

# Machine Learning
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, mean_absolute_percentage_error

# Deep Learning - LSTM Focus
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.optimizers import Adam

# Visualization - Professional Design
import matplotlib.patches as mpatches
from matplotlib.gridspec import GridSpec
import matplotlib.dates as mdates

# Set professional styling
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")


# ============================================================
# PHASE 1: DATA LOADING & PREPROCESSING
# ============================================================

def load_and_preprocess_data(file_path):
    """
    Load, clean, and preprocess electricity consumption data
    Optimized for LSTM model training

    Args:
        file_path (str): Path to CSV file

    Returns:
        DataFrame: Cleaned and preprocessed data
    """
    print("[PHASE 1] Loading & Preprocessing Data...")

    # Load data
    df = pd.read_csv(file_path)

    # Rename columns for consistency
    df.rename(columns={
        'Start time UTC': 'timestamp',
        'Electricity consumption (MWh)': 'consumption'
    }, inplace=True)

    # Drop unnecessary column
    if 'End time UTC' in df.columns:
        df.drop('End time UTC', axis=1, inplace=True)

    # Parse timestamp
    df['timestamp'] = pd.to_datetime(df['timestamp'], format='%m/%d/%Y %H:%M', utc=True)

    # Sort by timestamp
    df = df.sort_values('timestamp').reset_index(drop=True)

    # Handle missing values
    initial_missing = df['consumption'].isnull().sum()
    if initial_missing > 0:
        df['consumption'].fillna(method='ffill', inplace=True)

    # Remove outliers (IQR method)
    Q1 = df['consumption'].quantile(0.25)
    Q3 = df['consumption'].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df = df[(df['consumption'] >= lower_bound) & (df['consumption'] <= upper_bound)].reset_index(drop=True)

    print(f"✓ Data loaded: {len(df)} records")
    print(f"  Date range: {df['timestamp'].min()} to {df['timestamp'].max()}")
    print(f"  Mean consumption: {df['consumption'].mean():.2f} MWh")
    print(f"  Std deviation: {df['consumption'].std():.2f} MWh")

    return df


# ============================================================
# PHASE 2: FEATURE ENGINEERING FOR LSTM
# ============================================================

def create_lstm_features(df):
    """
    Create temporal features optimized for LSTM
    Focus on patterns that LSTM can learn effectively

    Args:
        df (DataFrame): Input data

    Returns:
        DataFrame: Data with engineered features
    """
    print("\n[PHASE 2] Creating LSTM-Optimized Features...")

    # Temporal features
    df['hour'] = df['timestamp'].dt.hour
    df['day_of_week'] = df['timestamp'].dt.dayofweek
    df['month'] = df['timestamp'].dt.month
    df['day_of_year'] = df['timestamp'].dt.dayofyear

    # Cyclical encoding for hour (important for daily patterns)
    df['hour_sin'] = np.sin(2 * np.pi * df['hour'] / 24)
    df['hour_cos'] = np.cos(2 * np.pi * df['hour'] / 24)

    # Cyclical encoding for month (important for seasonal patterns)
    df['month_sin'] = np.sin(2 * np.pi * df['month'] / 12)
    df['month_cos'] = np.cos(2 * np.pi * df['month'] / 12)

    # Lag features (LSTM learns these patterns)
    df['lag_24'] = df['consumption'].shift(24)  # 24 hours ago (same hour previous day)
    df['lag_168'] = df['consumption'].shift(168)  # 1 week ago (same hour/day)

    # Rolling statistics
    df['rolling_mean_24'] = df['consumption'].rolling(window=24).mean()
    df['rolling_std_24'] = df['consumption'].rolling(window=24).std()

    # Drop NaN values
    df = df.dropna().reset_index(drop=True)

    print(f"✓ Features created: {df.shape[1]} total columns")

    return df


# ============================================================
# PHASE 3: PREPARE DATA FOR LSTM
# ============================================================

def prepare_lstm_data(df, lookback=24, train_split=0.7, val_split=0.15):
    """
    Prepare sequences for LSTM training

    Args:
        df (DataFrame): Data with features
        lookback (int): Number of hours to look back (24 = 1 day)
        train_split (float): Training data percentage
        val_split (float): Validation data percentage

    Returns:
        X_train, y_train, X_val, y_val, X_test, y_test, scaler, original_data
    """
    print(f"\n[PHASE 3] Preparing LSTM Data (lookback={lookback}h)...")

    # Extract consumption values
    consumption = df['consumption'].values.reshape(-1, 1)

    # Normalize data
    scaler = MinMaxScaler(feature_range=(0, 1))
    scaled_data = scaler.fit_transform(consumption)

    # Create sequences
    X, y = [], []
    for i in range(lookback, len(scaled_data)):
        X.append(scaled_data[i-lookback:i, 0])
        y.append(scaled_data[i, 0])

    X = np.array(X)
    y = np.array(y)

    # Split data chronologically (time series!)
    n = len(X)
    train_size = int(n * train_split)
    val_size = int(n * val_split)

    X_train = X[:train_size]
    y_train = y[:train_size]

    X_val = X[train_size:train_size+val_size]
    y_val = y[train_size:train_size+val_size]

    X_test = X[train_size+val_size:]
    y_test = y[train_size+val_size:]

    # Reshape for LSTM: [samples, timesteps, features]
    X_train = X_train.reshape((X_train.shape[0], X_train.shape[1], 1))
    X_val = X_val.reshape((X_val.shape[0], X_val.shape[1], 1))
    X_test = X_test.reshape((X_test.shape[0], X_test.shape[1], 1))

    print(f"✓ Data prepared:")
    print(f"  X_train: {X_train.shape}, y_train: {y_train.shape}")
    print(f"  X_val: {X_val.shape}, y_val: {y_val.shape}")
    print(f"  X_test: {X_test.shape}, y_test: {y_test.shape}")

    # Save original timestamps for later use
    test_start_idx = train_size + val_size + lookback
    test_timestamps = df['timestamp'].iloc[test_start_idx:test_start_idx+len(y_test)].values

    return X_train, y_train, X_val, y_val, X_test, y_test, scaler, test_timestamps


# ============================================================
# PHASE 4: BUILD & TRAIN LSTM MODEL
# ============================================================

def build_and_train_lstm(X_train, y_train, X_val, y_val, lookback=24, epochs=100):
    """
    Build optimized LSTM architecture for energy forecasting

    Architecture designed for:
    - Capturing daily patterns (24-hour seasonality)
    - Learning long-term dependencies
    - Preventing overfitting

    Args:
        X_train, y_train: Training data
        X_val, y_val: Validation data
        lookback (int): Input sequence length
        epochs (int): Training iterations

    Returns:
        model: Trained LSTM model
        history: Training history
    """
    print("\n[PHASE 4] Building LSTM Model...")

    # Build model architecture
    model = Sequential([
        # LSTM Layer 1: 128 units, learns temporal patterns
        LSTM(128, activation='relu', return_sequences=True,
             input_shape=(lookback, 1),
             recurrent_dropout=0.2),
        Dropout(0.2),

        # LSTM Layer 2: 64 units, learns complex dependencies
        LSTM(64, activation='relu', return_sequences=False,
             recurrent_dropout=0.2),
        Dropout(0.2),

        # Dense layers
        Dense(32, activation='relu'),
        Dropout(0.1),
        Dense(16, activation='relu'),

        # Output layer: Single prediction
        Dense(1)
    ])

    # Compile with optimized settings for time series
    model.compile(
        optimizer=Adam(learning_rate=0.001),
        loss='mse',
        metrics=['mae']
    )

    print("\nLSTM Architecture:")
    model.summary()

    # Train with early stopping
    early_stop = EarlyStopping(
        monitor='val_loss',
        patience=15,
        restore_best_weights=True,
        verbose=1
    )

    print("\nTraining LSTM Model...")
    history = model.fit(
        X_train, y_train,
        validation_data=(X_val, y_val),
        epochs=epochs,
        batch_size=32,
        callbacks=[early_stop],
        verbose=1
    )

    print("✓ Model training complete!")

    return model, history


# ============================================================
# PHASE 5: EVALUATE LSTM ON TEST SET
# ============================================================

def evaluate_lstm(model, X_test, y_test, scaler, test_timestamps):
    """
    Evaluate LSTM performance on test set

    Args:
        model: Trained LSTM model
        X_test, y_test: Test data
        scaler: MinMaxScaler for inverse transform
        test_timestamps: Timestamps for test set

    Returns:
        Predictions and metrics
    """
    print("\n[PHASE 5] Evaluating LSTM on Test Set...")

    # Make predictions
    y_pred_scaled = model.predict(X_test, verbose=0)

    # Inverse transform to original scale
    y_pred = scaler.inverse_transform(y_pred_scaled)[:, 0]
    y_test_original = scaler.inverse_transform(y_test.reshape(-1, 1))[:, 0]

    # Calculate metrics
    mae = mean_absolute_error(y_test_original, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test_original, y_pred))
    mape = mean_absolute_percentage_error(y_test_original, y_pred)

    print(f"\n🎯 LSTM Test Set Performance:")
    print(f"  MAE:  {mae:.2f} MWh")
    print(f"  RMSE: {rmse:.2f} MWh")
    print(f"  MAPE: {mape:.2f}%")
    print(f"  Mean Consumption (Test): {y_test_original.mean():.2f} MWh")
    print(f"  Error as % of Mean: {(mae/y_test_original.mean())*100:.2f}%")

    return y_pred, y_test_original, mae, rmse, mape, test_timestamps


# ============================================================
# PHASE 6: FORECAST FUTURE ENERGY DEMAND (NEXT 7 DAYS)
# ============================================================

def forecast_future_demand(model, df, scaler, lookback=24, days_ahead=7):
    """
    Forecast future hourly energy demand using LSTM

    This function predicts the NEXT 7 days (168 hours) of energy consumption
    using the trained LSTM model in a recursive/autoregressive manner

    Args:
        model: Trained LSTM model
        df: Historical data with features
        scaler: MinMaxScaler for inverse transform
        lookback (int): Number of hours to look back
        days_ahead (int): Number of days to forecast

    Returns:
        forecast_df: DataFrame with future predictions and analysis
    """
    print(f"\n[PHASE 6] Forecasting Future Energy Demand ({days_ahead} days ahead)...")

    # Get last 'lookback' values from training data
    last_consumption = df['consumption'].values[-lookback:]
    last_timestamp = df['timestamp'].iloc[-1]

    # Normalize last values
    last_scaled = scaler.transform(last_consumption.reshape(-1, 1))

    # Generate forecast hours
    hours_to_forecast = days_ahead * 24
    forecast_timestamps = [last_timestamp + timedelta(hours=i) for i in range(1, hours_to_forecast + 1)]

    # Initialize forecast list
    forecast_values_scaled = []
    current_sequence = last_scaled.copy()

    # Recursive forecasting (autoregressive)
    print(f"  Generating {hours_to_forecast} hour-ahead predictions...")

    for hour in range(hours_to_forecast):
        # Reshape for LSTM: [1 sample, lookback timesteps, 1 feature]
        X_forecast = current_sequence.reshape(1, lookback, 1)

        # Predict next hour
        next_pred_scaled = model.predict(X_forecast, verbose=0)[0, 0]
        forecast_values_scaled.append(next_pred_scaled)

        # Update sequence: remove first value, add prediction
        current_sequence = np.vstack([current_sequence[1:], [[next_pred_scaled]]])

        # Progress indicator
        if (hour + 1) % 24 == 0:
            print(f"    ✓ Day {(hour + 1) // 24} forecast complete ({hour + 1}/{hours_to_forecast} hours)")

    # Inverse transform to original scale
    forecast_values_scaled = np.array(forecast_values_scaled).reshape(-1, 1)
    forecast_values = scaler.inverse_transform(forecast_values_scaled)[:, 0]

    # Create forecast DataFrame
    forecast_df = pd.DataFrame({
        'timestamp': forecast_timestamps,
        'predicted_consumption': forecast_values,
        'hour': [ts.hour for ts in forecast_timestamps],
        'day_of_week': [ts.day_name() for ts in forecast_timestamps],
        'day': [ts.strftime('%A, %b %d') for ts in forecast_timestamps]
    })

    # Add statistics
    forecast_df['rolling_mean_24h'] = forecast_df['predicted_consumption'].rolling(24).mean()

    print(f"✓ Forecast complete!")
    print(f"  Forecast period: {forecast_timestamps[0]} to {forecast_timestamps[-1]}")
    print(f"  Mean predicted demand: {forecast_values.mean():.2f} MWh")
    print(f"  Min predicted demand: {forecast_values.min():.2f} MWh")
    print(f"  Max predicted demand: {forecast_values.max():.2f} MWh")

    return forecast_df


# ============================================================
# PHASE 7: PROFESSIONAL VISUALIZATION - FUTURE FORECAST
# ============================================================

def visualize_future_forecast_professional(forecast_df, df):
    """
    Create professional, attention-grabbing visualization of future energy demand

    Design Principles:
    1. Clean, minimalist design
    2. Strategic use of color to draw attention
    3. Remove clutter - only essential elements
    4. Professional typography and spacing
    5. Focus on the forecast (main story)
    """
    print("\n[PHASE 7] Creating Professional Forecast Visualization...")

    # Define color palette - professional and meaningful
    COLOR_FORECAST = '#2E86AB'  # Professional blue for forecast
    COLOR_DAILY_AVG = '#A23B72'  # Accent color for daily average
    COLOR_CONFIDENCE = '#F18F01'  # Attention color for highlights

    # Create figure with professional styling
    fig = plt.figure(figsize=(20, 12))
    fig.patch.set_facecolor('white')

    # Create grid for complex layout
    gs = GridSpec(3, 2, figure=fig, hspace=0.35, wspace=0.25,
                  left=0.08, right=0.95, top=0.93, bottom=0.08)

    # ===== MAIN PLOT 1: HOURLY FORECAST (LARGEST, MOST PROMINENT) =====
    ax_main = fig.add_subplot(gs[0:2, 0:2])

    # Plot forecast line - thick and prominent
    ax_main.plot(forecast_df['timestamp'], forecast_df['predicted_consumption'],
                linewidth=3, color=COLOR_FORECAST, label='Predicted Hourly Demand',
                zorder=3, marker='o', markersize=4, markerfacecolor=COLOR_FORECAST,
                markeredgecolor='white', markeredgewidth=0.5)

    # Add daily average line - subtle but visible
    ax_main.plot(forecast_df['timestamp'], forecast_df['rolling_mean_24h'],
                linewidth=2.5, color=COLOR_DAILY_AVG, label='24-Hour Rolling Average',
                linestyle='--', zorder=2, alpha=0.8)

    # Fill under forecast for visual emphasis
    ax_main.fill_between(forecast_df['timestamp'],
                         forecast_df['predicted_consumption'],
                         alpha=0.15, color=COLOR_FORECAST, zorder=1)

    # Highlight peak demand hours
    max_demand_idx = forecast_df['predicted_consumption'].idxmax()
    max_demand_time = forecast_df.loc[max_demand_idx, 'timestamp']
    max_demand_value = forecast_df.loc[max_demand_idx, 'predicted_consumption']

    ax_main.scatter([max_demand_time], [max_demand_value],
                   s=400, color=COLOR_CONFIDENCE, marker='*',
                   zorder=5, edgecolor='white', linewidth=2,
                   label=f'Peak Demand: {max_demand_value:.0f} MWh')

    # Add annotation for peak
    ax_main.annotate(f'Peak: {max_demand_value:.0f} MWh\n{max_demand_time.strftime("%a, %H:%M")}',
                    xy=(max_demand_time, max_demand_value),
                    xytext=(20, 30), textcoords='offset points',
                    fontsize=11, fontweight='bold',
                    bbox=dict(boxstyle='round,pad=0.8', facecolor=COLOR_CONFIDENCE, alpha=0.9, edgecolor='white', linewidth=2),
                    arrowprops=dict(arrowstyle='->', color=COLOR_CONFIDENCE, lw=2.5, connectionstyle='arc3,rad=0.3'),
                    color='white', zorder=6)

    # Styling
    ax_main.set_xlabel('Date & Time', fontsize=13, fontweight='bold', labelpad=10)
    ax_main.set_ylabel('Energy Demand (MWh)', fontsize=13, fontweight='bold', labelpad=10)
    ax_main.set_title('7-Day Energy Demand Forecast (Hourly)',
                     fontsize=16, fontweight='bold', pad=20)

    # Format x-axis for readability
    ax_main.xaxis.set_major_locator(mdates.DayLocator())
    ax_main.xaxis.set_major_formatter(mdates.DateFormatter('%a\n%b %d'))
    plt.setp(ax_main.xaxis.get_majorticklabels(), fontsize=10, fontweight='bold')
    plt.setp(ax_main.yaxis.get_majorticklabels(), fontsize=10)

    # Grid - subtle
    ax_main.grid(True, alpha=0.2, linestyle='--', linewidth=0.5)
    ax_main.set_axisbelow(True)

    # Legend - professional placement
    ax_main.legend(loc='upper left', fontsize=11, framealpha=0.95,
                  edgecolor='lightgray', fancybox=True, shadow=True)

    # ===== SUBPLOT 2: HOURLY PATTERN BY DAY =====
    ax_hourly = fig.add_subplot(gs[2, 0])

    # Group by hour and calculate mean
    hourly_pattern = forecast_df.groupby('hour')['predicted_consumption'].mean()

    # Create bar chart with gradient colors
    bars = ax_hourly.bar(hourly_pattern.index, hourly_pattern.values,
                        color=COLOR_FORECAST, alpha=0.7, edgecolor='white', linewidth=1.5)

    # Highlight peak hours
    max_hour_idx = hourly_pattern.idxmax()
    bars[max_hour_idx].set_color(COLOR_CONFIDENCE)
    bars[max_hour_idx].set_alpha(1.0)

    ax_hourly.set_xlabel('Hour of Day', fontsize=11, fontweight='bold')
    ax_hourly.set_ylabel('Avg Demand (MWh)', fontsize=11, fontweight='bold')
    ax_hourly.set_title('Average Demand by Hour', fontsize=12, fontweight='bold')
    ax_hourly.set_xticks(range(0, 24, 2))
    ax_hourly.grid(True, alpha=0.2, axis='y')
    ax_hourly.set_axisbelow(True)

    # ===== SUBPLOT 3: DAILY DEMAND TOTALS =====
    ax_daily = fig.add_subplot(gs[2, 1])

    # Calculate daily totals
    daily_totals = forecast_df.groupby('day')['predicted_consumption'].sum()

    # Create area chart
    ax_daily.fill_between(range(len(daily_totals)), daily_totals.values,
                         color=COLOR_DAILY_AVG, alpha=0.5, edgecolor=COLOR_DAILY_AVG, linewidth=2.5)
    ax_daily.plot(range(len(daily_totals)), daily_totals.values,
                 color=COLOR_DAILY_AVG, linewidth=2.5, marker='o', markersize=8,
                 markerfacecolor='white', markeredgecolor=COLOR_DAILY_AVG, markeredgewidth=2)

    ax_daily.set_xlabel('Date', fontsize=11, fontweight='bold')
    ax_daily.set_ylabel('Daily Total (MWh)', fontsize=11, fontweight='bold')
    ax_daily.set_title('Daily Energy Demand', fontsize=12, fontweight='bold')
    ax_daily.set_xticks(range(len(daily_totals)))
    ax_daily.set_xticklabels([d.split(',')[0] for d in daily_totals.index], rotation=45, ha='right', fontsize=9)
    ax_daily.grid(True, alpha=0.2, axis='y')
    ax_daily.set_axisbelow(True)

    # Add summary statistics box
    summary_text = f"""
    FORECAST SUMMARY (7 Days)
    ━━━━━━━━━━━━━━━━━━━━━━━━━
    Mean Demand:    {forecast_df['predicted_consumption'].mean():.1f} MWh
    Peak Demand:    {forecast_df['predicted_consumption'].max():.1f} MWh
    Min Demand:     {forecast_df['predicted_consumption'].min():.1f} MWh
    Total Energy:   {forecast_df['predicted_consumption'].sum():.0f} MWh
    """

    fig.text(0.95, 0.05, summary_text, fontsize=10,
            bbox=dict(boxstyle='round', facecolor='#f0f0f0', alpha=0.9, pad=1, edgecolor='gray'),
            family='monospace', verticalalignment='bottom', horizontalalignment='right')

    # Main title
    fig.suptitle('LSTM Energy Demand Forecast - Next 7 Days',
                fontsize=18, fontweight='bold', y=0.98)

    # Subtitle with timestamp
    fig.text(0.08, 0.96, f'Generated: {datetime.now().strftime("%Y-%m-%d %H:%M:%S")} UTC',
            fontsize=10, style='italic', color='gray')

    # Save figure
    plt.savefig('10_lstm_future_forecast_professional.png', dpi=300, bbox_inches='tight',
               facecolor='white', edgecolor='none')
    print("✓ Saved: 10_lstm_future_forecast_professional.png")

    plt.show()


# ============================================================
# PHASE 8: ADDITIONAL COMPARISON VISUALIZATION
# ============================================================

def visualize_test_vs_forecast(df, y_pred, y_test, test_timestamps, forecast_df):
    """
    Compare historical test performance with future forecast
    Shows model reliability through demonstrated accuracy
    """
    print("\n[PHASE 8] Creating Test vs Forecast Comparison...")

    fig, axes = plt.subplots(2, 1, figsize=(18, 10))
    fig.patch.set_facecolor('white')

    # Color scheme
    COLOR_ACTUAL = '#1F77B4'
    COLOR_PRED = '#FF7F0E'

    # ===== PLOT 1: TEST SET PERFORMANCE (Proof of Model Reliability) =====
    ax1 = axes[0]

    # Convert test_timestamps to datetime if needed
    test_ts = pd.to_datetime(test_timestamps) if not isinstance(test_timestamps[0], pd.Timestamp) else test_timestamps

    ax1.plot(test_ts, y_test, linewidth=2.5, color=COLOR_ACTUAL, label='Actual Consumption',
            zorder=2, alpha=0.9)
    ax1.plot(test_ts, y_pred, linewidth=2, color=COLOR_PRED, label='LSTM Prediction',
            linestyle='--', zorder=1, alpha=0.8)

    ax1.fill_between(test_ts, y_test, y_pred, alpha=0.15, color=COLOR_ACTUAL, zorder=0)

    ax1.set_ylabel('Energy Demand (MWh)', fontsize=12, fontweight='bold')
    ax1.set_title('Historical Test Set: Actual vs LSTM Predictions (Demonstrates Model Accuracy)',
                 fontsize=13, fontweight='bold', pad=15)
    ax1.grid(True, alpha=0.2)
    ax1.legend(loc='upper right', fontsize=11, framealpha=0.95)

    # ===== PLOT 2: FUTURE FORECAST =====
    ax2 = axes[1]

    COLOR_FORECAST = '#2E86AB'
    ax2.plot(forecast_df['timestamp'], forecast_df['predicted_consumption'],
            linewidth=2.5, color=COLOR_FORECAST, label='7-Day Forecast', zorder=2, marker='o', markersize=3)
    ax2.fill_between(forecast_df['timestamp'], forecast_df['predicted_consumption'],
                    alpha=0.15, color=COLOR_FORECAST, zorder=1)

    ax2.set_xlabel('Date & Time', fontsize=12, fontweight='bold')
    ax2.set_ylabel('Energy Demand (MWh)', fontsize=12, fontweight='bold')
    ax2.set_title('Future Forecast: Next 7 Days of Hourly Energy Demand',
                 fontsize=13, fontweight='bold', pad=15)
    ax2.grid(True, alpha=0.2)
    ax2.legend(loc='upper right', fontsize=11, framealpha=0.95)

    # Format dates
    ax2.xaxis.set_major_locator(mdates.DayLocator())
    ax2.xaxis.set_major_formatter(mdates.DateFormatter('%a\n%b %d'))

    plt.tight_layout()
    plt.savefig('11_test_performance_vs_future_forecast.png', dpi=300, bbox_inches='tight',
               facecolor='white')
    print("✓ Saved: 11_test_performance_vs_future_forecast.png")

    plt.show()


# ============================================================
# PHASE 9: GENERATE FORECAST REPORT
# ============================================================

def generate_forecast_report(forecast_df, y_pred, y_test, mae, rmse, mape):
    """
    Generate professional text report of forecast
    """
    print("\n" + "="*100)
    print(" "*25 + "LSTM ENERGY DEMAND FORECAST REPORT")
    print("="*100)

    print("\n📊 MODEL PERFORMANCE ON HISTORICAL TEST DATA:")
    print("-" * 100)
    print(f"  Mean Absolute Error (MAE):       {mae:.2f} MWh")
    print(f"  Root Mean Squared Error (RMSE):  {rmse:.2f} MWh")
    print(f"  Mean Absolute Percentage Error:  {mape:.2f}%")
    print(f"  Model Accuracy:                  {(100-mape):.1f}%")

    print("\n📈 7-DAY FORECAST SUMMARY:")
    print("-" * 100)

    # By day statistics
    daily_stats = forecast_df.groupby('day')['predicted_consumption'].agg(['min', 'mean', 'max', 'sum'])

    print(f"\n{'Date':<20} {'Min (MWh)':<15} {'Mean (MWh)':<15} {'Max (MWh)':<15} {'Total (MWh)':<15}")
    print("-" * 100)
    for day, row in daily_stats.iterrows():
        print(f"{day:<20} {row['min']:<15.1f} {row['mean']:<15.1f} {row['max']:<15.1f} {row['sum']:<15.1f}")

    print("\n🎯 FORECAST INSIGHTS:")
    print("-" * 100)

    # Peak demand
    peak_idx = forecast_df['predicted_consumption'].idxmax()
    peak_time = forecast_df.loc[peak_idx, 'timestamp']
    peak_value = forecast_df.loc[peak_idx, 'predicted_consumption']

    # Low demand
    low_idx = forecast_df['predicted_consumption'].idxmin()
    low_time = forecast_df.loc[low_idx, 'timestamp']
    low_value = forecast_df.loc[low_idx, 'predicted_consumption']

    print(f"  Peak Demand:   {peak_value:.1f} MWh on {peak_time.strftime('%A, %b %d at %H:%M UTC')}")
    print(f"  Min Demand:    {low_value:.1f} MWh on {low_time.strftime('%A, %b %d at %H:%M UTC')}")
    print(f"  Average:       {forecast_df['predicted_consumption'].mean():.1f} MWh")
    print(f"  7-Day Total:   {forecast_df['predicted_consumption'].sum():.0f} MWh")

    # Daily pattern
    hourly_demand = forecast_df.groupby('hour')['predicted_consumption'].mean()
    peak_hour = hourly_demand.idxmax()

    print(f"\n  Peak Hour of Day:  {peak_hour:02d}:00 UTC")
    print(f"  Demand variation:  {(forecast_df['predicted_consumption'].max() - forecast_df['predicted_consumption'].min()):.1f} MWh")

    print("\n" + "="*100)


# ============================================================
# MAIN EXECUTION
# ============================================================

def main():
    """
    Main pipeline execution - LSTM focused energy demand forecasting
    """

    print("\n" + "="*100)
    print(" "*20 + "LSTM-BASED HOURLY ENERGY DEMAND FORECASTING PIPELINE")
    print(" "*20 + "Data-Driven Future Energy Consumption Predictions")
    print("="*100)

    # Phase 1-2: Load and preprocess data
    df = load_and_preprocess_data('/content/Energy prediction.csv')  # ← UPDATE WITH YOUR FILE PATH
    df = create_lstm_features(df)

    # Phase 3: Prepare LSTM sequences
    X_train, y_train, X_val, y_val, X_test, y_test, scaler, test_timestamps = prepare_lstm_data(df, lookback=24)

    # Phase 4: Build and train LSTM
    model, history = build_and_train_lstm(X_train, y_train, X_val, y_val, epochs=100)

    # Phase 5: Evaluate on test set
    y_pred, y_test_original, mae, rmse, mape, test_ts = evaluate_lstm(model, X_test, y_test, scaler, test_timestamps)

    # Phase 6: Forecast future demand (7 days ahead)
    forecast_df = forecast_future_demand(model, df, scaler, lookback=24, days_ahead=7)

    # Phase 7: Professional visualization of future forecast
    visualize_future_forecast_professional(forecast_df, df)

    # Phase 8: Compare test performance with forecast
    visualize_test_vs_forecast(df, y_pred, y_test_original, test_ts, forecast_df)

    # Phase 9: Generate detailed report
    generate_forecast_report(forecast_df, y_pred, y_test_original, mae, rmse, mape)

    print("\n✅ PIPELINE COMPLETE!")
    print("\nGenerated Visualizations:")
    print("  • 10_lstm_future_forecast_professional.png - Main 7-day forecast (MAIN DELIVERABLE)")
    print("  • 11_test_performance_vs_future_forecast.png - Test accuracy vs future forecast")

    return model, scaler, forecast_df


# ============================================================
# EXECUTION
# ============================================================

if __name__ == "__main__":
    model, scaler, forecast_df = main()
```
